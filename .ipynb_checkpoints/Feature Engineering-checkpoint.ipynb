```python
import pandas as pd
```

```python
df = pd.read_csv("Cleaned_dataset.csv")
df
```

```python
display(df.columns)
```

```python
df_copy = df.copy()
```

Converting categorical to numeric (Label encoding)

```python
from sklearn.preprocessing import LabelEncoder

le_gender = LabelEncoder()
df_copy['Gender'] = le_gender.fit_transform(df_copy['Gender']) # 1-Male, 0-Female

le_customer = LabelEncoder()
df_copy['Customer Type'] = le_customer.fit_transform(df_copy['Customer Type'])#First-time:0, Returning:1

le_travel = LabelEncoder()
df_copy['Type of Travel'] = le_travel.fit_transform(df_copy['Type of Travel'])#Business:0, 

le_class = LabelEncoder()
df_copy['Class'] = le_travel.fit_transform(df_copy['Class'])
```

```python
df_copy.head(10)
```

Age groups- categporize the age group as youth, adult, senior, elderly

```python
df_copy['Age Group'] = pd.cut(
    df_copy['Age'],
    bins=[0,25,45,65,100],
    labels=['Youth','Adult','Senior','Elderly']
)
display(df_copy)
```

```python
df_copy = pd.get_dummies(df_copy, columns=['Age Group'], drop_first=True)
display(df_copy)
```

Create a New Column "Total Service Score" of each users

```python
service_col = [
    'Departure and Arrival Time Convenience', 'Ease of Online Booking', 
    'Check-in Service', 'Online Boarding', 'Gate Location', 'On-board Service', 
    'Seat Comfort', 'Leg Room Service', 'Cleanliness', 'Food and Drink', 'In-flight Service', 
    'In-flight Wifi Service', 'In-flight Entertainment', 'Baggage Handling'
]

df_copy['Total Service Score'] = df_copy[service_col].sum(axis=1)
```

```python
display(df_copy)
```

```python
from sklearn.preprocessing import StandardScaler

num_cols = ['Age', 'Flight Distance', 'Departure Delay', 'Arrival Delay']

scaler = StandardScaler()
df_copy[num_cols] = scaler.fit_transform(df_copy[num_cols])
```

```python
display(df_copy)
```

```python
display("Mean after Normalization:\n", df_copy[num_cols].mean())
display("\nStandard Deviation after Normalization:\n", df_copy[num_cols].std())
```

```python
df_copy.to_csv("Feature_engineering.csv")
```
