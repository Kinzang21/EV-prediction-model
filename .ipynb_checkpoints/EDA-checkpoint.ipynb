```python
import pandas as pd
df=pd.read_csv("Cleaned_dataset .csv")
df.head()
```

```python
df.info()
```

```python
df.isnull().sum()
```

```python
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
satisfaction_counts = df["Satisfaction"].value_counts()
colors = ["#FFDE21", "#430059"]
mylabels=['Satisfied','Neutral or Dissatisfied']
plt.figure(figsize = (7,7))
plt.pie(satisfaction_counts, labels=mylabels,autopct="%1.1f%%", colors=colors)
plt.legend(title='Satisfaction',loc='upper right')
plt.title("Customer Satisfaction Distribution")
plt.show()
```

```python
satisfaction_by_gender = df.groupby(["Gender", "Satisfaction"]).size().unstack()

# Plot bar graph
satisfaction_by_gender.plot(kind="bar", figsize=(8, 5), colormap="viridis")

# Add labels and title
plt.xlabel("Gender")
plt.ylabel("Count of Passengers")
plt.title("Satisfaction Rates by Gender")
plt.xticks(rotation=0)  # Keep labels horizontal
plt.legend(title="Satisfaction Level")

# Show the plot
plt.show()
```

```python
# Count satisfaction levels for each customer type
satisfaction_by_customer = df.groupby("Customer Type")["Satisfaction"].value_counts().unstack()

# Plot stacked bar chart
satisfaction_by_customer.plot(kind="bar", stacked=False, figsize=(8, 6), colormap="viridis")

# Add labels and title
plt.xlabel("Customer Type")
plt.ylabel("Number of Passengers")
plt.title("Satisfaction Levels by Customer Type")
plt.legend(title="Satisfaction Level")
plt.xticks(rotation=0)  # Keep labels readable
plt.show()
```

```python
# Count satisfaction levels for each type of travel
satisfaction_by_travel = df.groupby("Type of Travel")["Satisfaction"].value_counts().unstack()

# Plot stacked bar chart
satisfaction_by_travel.plot(kind="bar", figsize=(8, 6), colormap="viridis")

# Add labels and title
plt.xlabel("Type of Travel")
plt.ylabel("Number of Passengers")
plt.title("Satisfaction Levels by Type of Travel")
plt.legend(title="Satisfaction Level")
plt.xticks(rotation=0)  # Keep labels readable
plt.show()
```

```python
# Count satisfaction levels for each class
satisfaction_by_class = df.groupby("Class")["Satisfaction"].value_counts().unstack()

# Plot stacked bar chart
satisfaction_by_class.plot(kind="bar", figsize=(8, 6), colormap="viridis")

# Add labels and title
plt.xlabel("Class")
plt.ylabel("Number of Passengers")
plt.title("Satisfaction Levels by Travel Class")
plt.legend(title="Satisfaction Level")
plt.xticks(rotation=0)  # Keep class labels readable
plt.show()
```

```python
plt.figure(figsize=(8,6))
sns.scatterplot(data=df, x="Flight Distance", y="Satisfaction", hue="Satisfaction", palette="viridis", alpha=0.7)

# Add labels and title
plt.xlabel("Flight Distance")
plt.ylabel("Satisfaction Level")
plt.title("Scatter Plot: Flight Distance vs. Satisfaction")

plt.show()
```

```python
df.groupby("Satisfaction")["Departure Delay"].mean().plot(kind="bar", color="#430059", edgecolor="black")
plt.ylabel("Departure Delay (Minutes)")
plt.show()
```

```python
df.groupby("Satisfaction")["Arrival Delay"].mean().plot(kind="bar", color="#430059", edgecolor="black")
plt.ylabel("Arrival Delay (Minutes)")
plt.show()
```

```python
df.groupby("Satisfaction")["Departure and Arrival Time Convenience"].mean().plot(kind="bar", color="#430059", edgecolor="black")
plt.ylabel("Departure and Arrival Time Convenience")
plt.show()
```

```python
# Calculate average leg room rating per satisfaction level
legroom_avg = df.groupby('Satisfaction')['Leg Room Service'].mean().reset_index()

# Plotting bar graph
plt.figure(figsize=(8, 5))
sns.barplot(x='Satisfaction', y='Leg Room Service', data=legroom_avg, color="#430059", edgecolor="black")
plt.title('Average Leg Room Service Rating by Satisfaction')
plt.xlabel('Satisfaction')
plt.ylabel('Average Leg Room Rating')

plt.show()
```
